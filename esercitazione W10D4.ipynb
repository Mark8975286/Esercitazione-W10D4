# Esercizio 1 · Valori nulli (semplice)

## Esercizio 1.1
**Domanda:** costruire il DataFrame `clienti` con le colonne nome, eta e citta per 8 righe, inserendo almeno tre valori nulli (`None`) in totale, non tutti sulla stessa colonna.

```python
dati_clienti = {
    "nome": ["Anna", "Luca", "Marta", None, "Paolo", "Giulia", "Davide", "Sara"],
    "eta": [34, None, 28, 45, 52, None, 39, 23],
    "citta": ["Torino", "Milano", None, "Roma", "Torino", "Napoli", "Milano", "Torino"],
}

clienti = pd.DataFrame(dati_clienti)

print(clienti)
print()
print("Shape:", clienti.shape)
print("Nulli inseriti -> nome: 1, eta: 2, citta: 1 (totale 4)")
```

## Esercizio 1.2
**Domanda:** usare `.isna()` sull'intero DataFrame e poi `.isna().sum()` per contare i valori nulli in ciascuna colonna.

```python
print("=== clienti.isna() ===")
print(clienti.isna())
```

```python
print("=== clienti.isna().sum() ===")
print(clienti.isna().sum())
print()
print("Totale valori nulli:", clienti.isna().sum().sum())
```

**Verifica:** il conteggio per colonna (nome 1, eta 2, citta 1) corrisponde ai `None` inseriti al punto 1.

## Esercizio 1.3
**Domanda:** usare `.dropna()` per ottenere un nuovo DataFrame senza righe con valori nulli, assegnando il risultato a una nuova variabile.

```python
clienti_puliti = clienti.dropna()

print(clienti_puliti)
print()
print("Righe DataFrame originale:", len(clienti))
print("Righe dopo dropna():      ", len(clienti_puliti))
print("Originale invariato? Nulli ancora presenti:", clienti.isna().sum().sum())
```

### Definizione richiesta
**Valore nullo in Pandas:** osservazione mancante in una cella, rappresentata da `NaN`, `NaT` (per le date) o `None`, e rilevata con `.isna()`.

**Perché `.dropna()` non modifica l'originale:** `.dropna()` non lavora in loco ma **restituisce un nuovo DataFrame** con le righe filtrate; `clienti` resta invariato (8 righe) e il risultato si conserva solo assegnandolo a una variabile, qui `clienti_puliti`.

# Esercizio 2 · Ordinamento (semplice)

## Esercizio 2.1
**Domanda:** ordinare il DataFrame `prodotti` per la colonna prezzo in ordine decrescente con `.sort_values()`, assegnando il risultato a una nuova variabile.

```python
prodotti = pd.DataFrame({
    "nome": ["Penna", "Zaino", "Quaderno", "Mouse", "Cuffie", "Evidenziatore"],
    "categoria": ["Cancelleria", "Accessori", "Cancelleria", "Elettronica", "Elettronica", "Cancelleria"],
    "prezzo": [1.50, 25.00, 3.00, 15.90, 49.90, 2.20],
})
print(prodotti)
print()

prodotti_per_prezzo = prodotti.sort_values("prezzo", ascending=False)

print("=== Ordinati per prezzo decrescente ===")
print(prodotti_per_prezzo)
```

## Esercizio 2.2
**Domanda:** ordinare il DataFrame passando una lista di due colonne a `.sort_values()`: prima categoria in ordine crescente, poi prezzo in ordine decrescente.

```python
prodotti_cat_prezzo = prodotti.sort_values(["categoria", "prezzo"], ascending=[True, False])

print("=== Ordinati per categoria (A→Z) e poi prezzo (decrescente) ===")
print(prodotti_cat_prezzo)
```

## Esercizio 2.3
**Domanda:** confrontare le prime due righe dei risultati dei punti 1 e 2 e spiegare perché sono diverse.

```python
print("Prime due righe – punto 1:")
print(prodotti_per_prezzo.head(2))
print()
print("Prime due righe – punto 2:")
print(prodotti_cat_prezzo.head(2))
```

**Motivazione:** nel punto 1 il criterio unico è il prezzo, quindi in cima compaiono i due prodotti più cari in assoluto (Cuffie e Zaino); nel punto 2 il primo criterio è la categoria in ordine alfabetico, quindi in cima va "Accessori" (Zaino) e il prezzo decide l'ordine solo all'interno di ciascuna categoria, perciò la seconda riga è il prodotto più caro di "Cancelleria" (Quaderno).

### Definizione richiesta
**Ordinamento su più colonne:** ordinamento in cui le righe vengono disposte secondo la prima colonna indicata e, a parità di valore, secondo la colonna successiva.

**Parametro `ascending` con una lista:** permette di indicare una direzione diversa per ciascuna colonna, nello stesso ordine della lista di colonne: `ascending=[True, False]` significa categoria crescente e prezzo decrescente.

# Esercizio 3 · Filtri con condizioni multiple (semplice)

## Esercizio 3.1
**Domanda:** costruire un filtro che selezioni i prodotti con prezzo superiore a una soglia a scelta e categoria uguale a un valore a scelta, combinando le due condizioni con l'operatore `&`.

```python
soglia = 10
categoria_scelta = "Elettronica"

filtro_and = (prodotti["prezzo"] > soglia) & (prodotti["categoria"] == categoria_scelta)
risultato_and = prodotti[filtro_and]

print(f"Prodotti con prezzo > {soglia} € E categoria = {categoria_scelta}:", len(risultato_and), "righe")
print(risultato_and)
```

## Esercizio 3.2
**Domanda:** usare `.isin()` per selezionare i prodotti che appartengono a una lista di due categorie a scelta, senza scrivere due condizioni unite da `|`.

```python
categorie_scelte = ["Cancelleria", "Accessori"]

filtro_isin = prodotti["categoria"].isin(categorie_scelte)
risultato_isin = prodotti[filtro_isin]

print(f"Prodotti nelle categorie {categorie_scelte}:", len(risultato_isin), "righe")
print(risultato_isin)
```

## Esercizio 3.3
**Domanda:** negare il filtro del punto 2 con l'operatore `~` per ottenere i prodotti che non appartengono a quelle categorie.

```python
risultato_not = prodotti[~filtro_isin]

print(f"Prodotti NON nelle categorie {categorie_scelte}:", len(risultato_not), "righe")
print(risultato_not)
print()
print("Righe .isin():", len(risultato_isin), "+ righe ~:", len(risultato_not),
      "=", len(risultato_isin) + len(risultato_not), "| Righe totali:", len(prodotti))
print("La somma corrisponde al totale?", len(risultato_isin) + len(risultato_not) == len(prodotti))
```

**Nota:** `&` e `|` hanno priorità maggiore degli operatori di confronto, quindi ogni condizione combinata va racchiusa tra parentesi, come in `(df["a"] > 1) & (df["b"] == "x")`.

### Definizione richiesta
**`.isin()`:** metodo che restituisce una maschera booleana con `True` per le righe il cui valore è contenuto nella lista passata; sostituisce più condizioni `==` unite da `|`.

**Perché si usano `&` e `|` invece di `and` e `or`:** `and` e `or` valutano l'intera Series come un unico `True` o `False` e generano un `ValueError`; `&` e `|` invece confrontano le maschere **elemento per elemento**, riga per riga.

# Esercizio 4 · Strategie di imputazione (avanzato)

## Esercizio 4.1
**Domanda:** applicare separatamente `.fillna()` con un valore fisso a scelta, `.ffill()` e `.interpolate()` alla stessa Series, ottenendo tre risultati distinti.

```python
temperature = pd.Series([20, 21, None, 24, 25, 23])
print("Series originale:")
print(temperature.tolist(), "| dtype:", temperature.dtype)
print()

imput_fillna = temperature.fillna(0)
imput_ffill = temperature.ffill()
imput_interp = temperature.interpolate()

print("fillna(0):     ", imput_fillna.tolist())
print("ffill():       ", imput_ffill.tolist())
print("interpolate(): ", imput_interp.tolist())
```

## Esercizio 4.2
**Domanda:** applicare anche `.bfill()` alla stessa Series e confrontare il suo risultato con quello di `.ffill()` sul valore che era nullo.

```python
imput_bfill = temperature.bfill()

print("bfill():", imput_bfill.tolist())
print()
posizione_nulla = temperature[temperature.isna()].index[0]
print("Posizione del valore nullo:", posizione_nulla)
print("Valore imputato con ffill():", imput_ffill[posizione_nulla], "(copia il valore PRECEDENTE)")
print("Valore imputato con bfill():", imput_bfill[posizione_nulla], "(copia il valore SUCCESSIVO)")
```

```python
# Riepilogo dei quattro metodi applicati alla stessa Series originale
confronto = pd.DataFrame({
    "originale": temperature,
    "fillna(0)": imput_fillna,
    "ffill": imput_ffill,
    "bfill": imput_bfill,
    "interpolate": imput_interp,
})
print(confronto)
```

## Esercizio 4.3
**Domanda:** indicare, per un caso in cui il valore nullo fosse il primo elemento della Series invece che uno intermedio, quale dei quattro metodi non riuscirebbe a riempirlo, motivando la risposta.

```python
temperature_inizio = pd.Series([None, 21, 22, 24, 25, 23])

print("Originale:     ", temperature_inizio.tolist())
print("fillna(0):     ", temperature_inizio.fillna(0).tolist())
print("ffill():       ", temperature_inizio.ffill().tolist())
print("bfill():       ", temperature_inizio.bfill().tolist())
print("interpolate(): ", temperature_inizio.interpolate().tolist())
```

**Risposta:** il metodo che non riesce a riempirlo è **`.ffill()`**, perché copia il valore **precedente** e prima del primo elemento non esiste nessun valore da propagare, quindi il `NaN` resta. Come mostra l'output, anche `.interpolate()` (lineare, impostazione predefinita) lo lascia vuoto, perché per stimare un valore ha bisogno di un dato prima e di uno dopo. `.bfill()` e `.fillna()` invece lo riempiono.

### Definizione richiesta
**Imputazione:** sostituzione di un valore nullo con un valore fisso, statistico o stimato dai valori circostanti.

**Differenza tra `.fillna()` e `.interpolate()`:** con `.fillna()` il valore imputato è **scelto da chi analizza** (una costante o una statistica come la media) ed è lo stesso a prescindere dalla posizione; con `.interpolate()` il valore è **stimato dai dati vicini** (qui la media tra 21 e 24 = 22.5), quindi segue l'andamento della serie.

# Esercizio 5 · Raggruppamento e aggregazione (avanzato)

## Esercizio 5.1
**Domanda:** raggruppare il DataFrame `vendite` per la colonna prodotto e calcolare la media della colonna importo per ciascun gruppo con `.groupby().mean()`.

```python
vendite = pd.DataFrame({
    "prodotto": ["Pane", "Pane", "Pane", "Pane", "Torta", "Torta",
                 "Torta", "Torta", "Biscotti", "Biscotti", "Biscotti", "Biscotti"],
    "regione": ["Piemonte", "Piemonte", "Lombardia", "Lombardia", "Piemonte", "Piemonte",
                "Lombardia", "Lombardia", "Piemonte", "Piemonte", "Lombardia", "Lombardia"],
    "importo": [120, 150, 200, 180, 300, 340, 410, 390, 80, 95, 110, 105],
})
print(vendite)
print()

media_per_prodotto = vendite.groupby("prodotto")["importo"].mean()

print("=== Media importo per prodotto ===")
print(media_per_prodotto)
```

## Esercizio 5.2
**Domanda:** usare `.agg()` per calcolare contemporaneamente media e somma di importo per ciascun prodotto, in un'unica istruzione.

```python
media_somma = vendite.groupby("prodotto")["importo"].agg(["mean", "sum"])

print("=== Media e somma per prodotto ===")
print(media_somma)
```

## Esercizio 5.3
**Domanda:** raggruppare per le due colonne prodotto e regione insieme, e usare `.get_group()` per recuperare le righe originali di una combinazione specifica.

```python
gruppi = vendite.groupby(["prodotto", "regione"])

print("Combinazioni presenti:", list(gruppi.groups.keys()))
print()
print("=== Righe originali di (Torta, Piemonte) ===")
print(gruppi.get_group(("Torta", "Piemonte")))
```

```python
# Il solo groupby, senza aggregazione, non mostra i dati
print(vendite.groupby("prodotto"))
print(type(vendite.groupby("prodotto")))
```

### Definizione richiesta
**Oggetto GroupBy:** oggetto intermedio creato da `.groupby()` che contiene la suddivisione delle righe in gruppi, ma diventa un risultato leggibile solo dopo un metodo di aggregazione.

**Perché `.groupby()` da solo non produce un output leggibile:** `.groupby()` si limita a stabilire *quali righe stanno insieme* senza calcolare nulla, quindi stampandolo si vede solo l'indirizzo dell'oggetto in memoria; serve un metodo come `.mean()`, `.sum()` o `.agg()` per ridurre ogni gruppo a un valore.

# Esercizio 6 · Merge, concat e date (avanzato)

## Esercizio 6.1
**Domanda:** unire `ordini` e `clienti` con `pd.merge()`, usando `id_cliente` come colonna chiave comune e `how="left"`, per ottenere il nome del cliente su ogni riga di ordine.

```python
ordini = pd.DataFrame({
    "id_ordine": [1, 2, 3, 4, 5],
    "id_cliente": [10, 20, 10, 30, 20],
    "importo": [120.0, 75.5, 210.0, 48.9, 99.0],
    "data_str": ["05/01/2026", "18/01/2026", "03/02/2026", "21/02/2026", "10/03/2026"],
})
anagrafica_clienti = pd.DataFrame({
    "id_cliente": [10, 20, 30],
    "nome": ["Rossi", "Bianchi", "Verdi"],
})

ordini_clienti = pd.merge(ordini, anagrafica_clienti, on="id_cliente", how="left")

print(ordini_clienti)
print()
print("Righe senza nome cliente:", ordini_clienti["nome"].isna().sum())
```

## Esercizio 6.2
**Domanda:** convertire la colonna `data_str` del DataFrame unito in un vero datatype data con `pd.to_datetime()`, usando il parametro `dayfirst` appropriato per il formato "gg/mm/aaaa", in una nuova colonna `data`.

```python
ordini_clienti["data"] = pd.to_datetime(ordini_clienti["data_str"], dayfirst=True)

# Con Pandas 3 la precisione predefinita può essere il microsecondo (datetime64[us]);
# la si porta esplicitamente a nanosecondi come richiesto dalla consegna.
ordini_clienti["data"] = ordini_clienti["data"].astype("datetime64[ns]")

print(ordini_clienti[["data_str", "data"]])
print()
print("dtype data_str:", ordini_clienti["data_str"].dtype)
print("dtype data:    ", ordini_clienti["data"].dtype)
```

## Esercizio 6.3
**Domanda:** usare l'accessor `.dt` sulla colonna `data` per estrarre il mese in una colonna aggiuntiva, poi usare `pd.concat()` per accodare un secondo piccolo DataFrame di ordini con le stesse colonne.

```python
ordini_clienti["mese"] = ordini_clienti["data"].dt.month
print(ordini_clienti)
```

```python
nuovi_ordini = pd.DataFrame({
    "id_ordine": [6, 7],
    "id_cliente": [30, 10],
    "importo": [155.0, 62.4],
    "data_str": ["02/04/2026", "15/04/2026"],
    "nome": ["Verdi", "Rossi"],
})
nuovi_ordini["data"] = pd.to_datetime(nuovi_ordini["data_str"], dayfirst=True).astype("datetime64[ns]")
nuovi_ordini["mese"] = nuovi_ordini["data"].dt.month

ordini_totali = pd.concat([ordini_clienti, nuovi_ordini], ignore_index=True)

print(ordini_totali)
print()
print("Righe primo DataFrame:", len(ordini_clienti), "+ righe secondo:", len(nuovi_ordini),
      "=", len(ordini_totali))
print("dtype data:", ordini_totali["data"].dtype)
```

### Definizione richiesta
**Merge:** operazione che unisce due DataFrame **affiancando le colonne** e abbinando le righe in base ai valori di una colonna chiave comune (qui `id_cliente`); `how="left"` conserva tutte le righe del DataFrame di sinistra.

**Perché `pd.concat()` non richiede una chiave comune:** `concat()` non abbina righe tra loro ma si limita ad **accodare** un DataFrame sotto l'altro (o accanto), allineando le colonne per nome; non deve capire quale riga corrisponde a quale, quindi non ha bisogno di una chiave, mentre `merge()` sì, perché deve sapere quali righe dei due DataFrame vanno unite.
